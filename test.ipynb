```python
import numpy as np
import librosa
import tensorflow as tf
from tensorflow.keras.models import load_model
```

```python
def test_saved_model(audio_path, model_path, max_length=500):
    # 1. Load the saved model
    model = load_model(model_path)
    
    # 2. Preprocess the test audio file
    audio, _ = librosa.load(audio_path, sr=16000)
    mfccs = librosa.feature.mfcc(y=audio, sr=16000, n_mfcc=40)
    
    # 3. Process features same as training
    if mfccs.shape[1] < max_length:
        mfccs = np.pad(mfccs, ((0, 0), (0, max_length - mfccs.shape[1])), mode='constant')
    else:
        mfccs = mfccs[:, :max_length]
    
    # 4. Reshape for model input
    input_data = np.expand_dims(np.expand_dims(mfccs, axis=0), axis=-1)
    
    # 5. Make prediction
    prediction = model.predict(input_data)
    confidence = prediction[0][0]
    label = "Fake" if confidence > 0.5 else "Real"
    
    print(f"Prediction: {label} (Confidence: {confidence*100:.2f}%)")
    return label, confidence

# Example usage
model_path = 'baseline_model/Baseline_Model.h5'
test_audio = '/Users/aviralchauhan/College/Sem8/hcai/project/final/test_data/deepfake.wav'

test_saved_model(test_audio, model_path)
```
